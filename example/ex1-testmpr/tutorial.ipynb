# Processing test.mpr file

## Load required libs

```python
import sys
import os

# add project root (alpobat folder) to path
sys.path.append(os.path.abspath("../../../"))

from alpobat.lib import mprtool, plotdata

%matplotlib widget
```

```python
# mprtool  has tools: readmpr, splitdata, write_csv
# plotdata has tools: plot_Qchargedischarge, plot_bode_impedance, plot_timevsEandI, plot_cyclevsEandI, plot_timecyclevsEandI

# ---> More to be added as per need
```

## Read mpr file

```python
loc = os.getcwd()

ds = mprtool(path=os.path.join(loc), filename="test.mpr")
df = ds.load()
```

## Save selected data to file over selected cycles

```python
ds.export_csv(columns=["freq", "|Z|"], start_cycle=10, end_cycle=12)
```

## Plot $Q_{ave}$

```python
figQ, a = plotdata.plot_Qchargedischarge(df, showtime=True, grid=True, charge_color='k', discharge_color='r')
```

```python
figQ.savefig("qave.png", dpi=600)
```

## Plot Bode impedance

```python
plotdata.plot_bode_impedance(df, startcycle=10, endcycle=12)
```

## Plot $ time ~ vs ~ E ~ \& ~ I$ over selected cycle range (over time range not implemented)

```python
plotdata.timevsEandI(df, startcycle=5, endcycle=10, showlegend=False, savefig=False)
```

## Plot $ cycles ~ vs ~ E ~ \& ~ I$ over selected cycle range

```python
plotdata.cyclevsEandI(df, startcycle=5, endcycle=None, showlegend=False)
```

## Plot $ (time,~cycles) ~ vs ~ (E ~ \& ~ I)$ over selected cycle range

```python
plotdata.timecyclevsEandI(df, startcycle=5, endcycle=None, interval=10, savefig=False, show=True)
```
